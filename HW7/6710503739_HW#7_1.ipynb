```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

data = [
    ("Red",    "Sport", "Domestic", "Yes"),
    ("Red",    "Sport", "Domestic", "No"),
    ("Red",    "Sport", "Domestic", "Yes"),
    ("Yellow", "Sport", "Domestic", "No"),
    ("Yellow", "Sport", "Imported", "Yes"),
    ("Yellow", "SUV",   "Imported", "No"),
    ("Yellow", "SUV",   "Imported", "Yes"),
    ("Yellow", "SUV",   "Domestic", "No"),
    ("Red",    "SUV",   "Imported", "No"),
    ("Red",    "Sport", "Imported", "Yes"),
]
df = pd.DataFrame(data, columns=["Colour", "Type", "Origin", "Stolen"])
df
```

## 1. Encode the attributes

```python
X = pd.get_dummies(df[["Colour", "Type", "Origin"]]).astype(int)
y = df["Stolen"]

query = pd.DataFrame([{"Colour": "Red", "Type": "SUV", "Origin": "Domestic"}])
Xq = pd.get_dummies(query).reindex(columns=X.columns, fill_value=0).astype(int)

X.head()
```

## 2. Train KNN and predict

k = 5

```python
knn = KNeighborsClassifier(n_neighbors=5, metric="euclidean")
knn.fit(X, y)

pred = knn.predict(Xq)[0]
proba = knn.predict_proba(Xq)[0]

print("Prediction for (Red, SUV, Domestic):", pred)
```

## 3. Which neighbours produced that vote?

```python
dist, idx = knn.kneighbors(Xq, n_neighbors=5)
neighbours = df.iloc[idx[0]].copy()
neighbours["Distance"] = dist[0].round(3)
neighbours
```

## Answer

Their majority label is No, so the KNN prediction for
(Red, SUV, Domestic) is Stolen = No.

the *k* = 5 result is the one that actually reflects the data.
